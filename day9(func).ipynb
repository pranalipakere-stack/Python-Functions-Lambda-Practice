```python
#Q2.Write a function that takes a number as a parameter and prints its square.
def sqr(n1):
    s2=n1**2
    return s2
total=sqr(5)
print(total)
```

```python
#Q3.Write a function that takes two numbers and prints their sum.
def add(n1,n2):
    sum=n1+n2
    return sum
total=add(20,40)
print(total)
```

```python
#Q4.Write a function to check whether a number is even or odd.
def even(n1):
    if n1%2==0:
        print("even number")
    else:
        print("odd number")
even(5) 
```

```python
#Q5.Write a function to find the largest of two numbers.
def lar(n1,n2):
    if n1>n2:
        print(n1)
    else:
        print(n2)
lar(50,100)
```

```python
#Q6. Write a function to calculate the area of a rectangle.
def rect(l,b):
    area=l*b
    return area
total=rect(9,5)
print(total)
```

```python
#Q7. Write a function to print the multiplication table of a given number.
def mul(n1):
    for i in range(1,11):
        print(n1*i)
mul(9)
```

```python
#Q8.Write a function to count the number of vowels in a string.
def vowel(data):
    count=0
    for i in data:
        if i=="i" or i=="o" or i=="a" or i=="e" or i=="u":
            count=count+1
    return count
total=vowel("pranali")
print(total)
```

```python
#Q9.Write a function to find the factorial of a number.
def fact(n1):
    fact1=1
    for i in range(1,n1+1):
        fact1=fact1*i
    return fact1
total=fact(7)
print(total)
```

```python
#Q10. Write a function to find the sum of all numbers in a list.
L1=[10,30,40,60,50]
def lis():
    sum=0
    for i in L1:
        sum=sum+i
    return sum 
total=lis()
print(total)
```
