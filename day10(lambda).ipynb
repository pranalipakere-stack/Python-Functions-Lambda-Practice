```python
#Q1.Create a lambda function to add two numbers.
add=lambda a,b:a+b
add(10,20)
```

```python
#Q2.Create a lambda function to find the square of a number.
sqr=lambda a:a**2
sqr(10)
```

```python
#Q3. Create a lambda function to find the cube of a number.
cube=lambda a:a**3
cube(2)
```

```python
#Q4. Create a lambda function to check whether a number is even or odd.
evodd=lambda a:"even" if a%2!=1 else "odd"
evodd(2)
```

```python
#Q5.Create a lambda function to find the larger of two numbers.
great=lambda a,b:a if a>b else b
great(20,10)
```

```python
#Q6.Create a lambda function to multiply three numbers.
multi=lambda a,b,c:a*b*c
multi(9,9,9)
```

```python
#Q7.Create a lambda function to calculate the area of a rectangle.
area=lambda l,b:l*b
area(9,6)
```

```python
#Q8.Use a lambda function with map() to square all elements in a list.
l=[1,2,3,4,5,6,7]
list(map(lambda x:x**2,l))
```

```python
#Q9.Use a lambda function with filter() to find all even numbers from a list.
list(filter(lambda x:x%2==0,l))
```

```python
#Q10.Use a lambda function with sorted() to sort a list of tuples based on the second element.
l1=[("A",50),("B",20),("C",40)]
sorted(l1,key=lambda x:x[1])
```
